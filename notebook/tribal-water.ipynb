```python
## Import the library tool into this VS Code software so we can analyse files 

import pandas as pd
```

```python
## Look specifically at the large EPA file exported with the ECHO Exporter 2.0
## LP, this step below is important. You just learned how to direct the code to a specific file. How cool!

path = "../data/copy/copy_ECHO_EXPORTER_copy.csv"

## Next we are dropping columns that are not relevant to our analysis. 
## We will keep the columns that are relevant to our analysis.
## Note, LP inputed this list into Claude for formatting. Then she compared the number of columns to remove based on her own list.  
## She confirmed it's the correct number of column names to remove.

drop = [
    "FAC_US_MEX_BORDER_FLG", "FAC_CHESAPEAKE_BAY_FLG", "FAC_COLLECTION_METHOD",
    "FAC_REFERENCE_POINT", "FAC_DERIVED_CD113", "FAC_DERIVED_CB2010", "FAC_MYRTK_UNIVERSE",
    "GHG_FLAG", "RCRA_FLAG", "AIR_FLAG", "AIR_IDS",
    "CAA_PERMIT_TYPES", "CAA_NAICS", "CAA_SICS", "CAA_EVALUATION_COUNT",
    "CAA_DAYS_LAST_EVALUATION", "CAA_INFORMAL_COUNT", "CAA_FORMAL_ACTION_COUNT",
    "CAA_DATE_LAST_FORMAL_ACTION", "CAA_PENALTIES", "CAA_LAST_PENALTY_DATE",
    "CAA_LAST_PENALTY_AMT", "CAA_QTRS_WITH_NC", "CAA_COMPLIANCE_STATUS",
    "CAA_HPV_FLAG", "CAA_3YR_COMPL_QTRS_HISTORY",
    "RCRA_IDS", "RCRA_PERMIT_TYPES", "RCRA_NAICS", "RCRA_INSPECTION_COUNT",
    "RCRA_DAYS_LAST_EVALUATION", "RCRA_INFORMAL_COUNT", "RCRA_FORMAL_ACTION_COUNT",
    "RCRA_DATE_LAST_FORMAL_ACTION", "RCRA_PENALTIES", "RCRA_LAST_PENALTY_DATE",
    "RCRA_LAST_PENALTY_AMT", "RCRA_QTRS_WITH_NC", "RCRA_COMPLIANCE_STATUS",
    "RCRA_SNC_FLAG", "RCRA_3YR_COMPL_QTRS_HISTORY",
    "GHG_IDS", "GHG_CO2_RELEASES",
]

## Next we are just reading the header of the file to see what columns are there. We are not reading any rows for now. 
## Because the file is so large, we are just reading the header to see what columns are there.

header = pd.read_csv(path, nrows=0).columns

## Now we are checking to see if any of the columns we want to drop are not in the header of the file. (I.e. We are taking a sec to /
## double check that the columns we want to drop are actually in the file. If they are not, we will print them out so we can see which ones are not in the file.)

print("Not found in file:", [c for c in drop if c not in header])
```

```python
## This not found in file empty result is good. It means that all the columns we want to drop are actually in the file.

## We are now creating a new datafram (spreadsheet) that will only have the columns we want to keep. 
## We are using the usecols parameter to specify which columns we want to keep.
## I think the "low_memory=False" parameter is used to tell pandas to not try to guess the data types of the columns.

df_echo = pd.read_csv(path, dtype=str, usecols=lambda c: c not in drop, low_memory=False)
```

```python
## Now we are double checking to see if the number of columns in the new dataframe is what we expect.

print(len(header), "->", len(df_echo.columns))   # the difference should be 43 because we removed 43 columns.
```

```python
## We nailed it. This is the number of columns we expected to have after dropping the columns we don't need.
```

```python
## Next we will check the number of rows and columns in the new dataframe. This will give us an idea of how many rows and columns we have to work with.

print(df_echo.shape)
```
